```python
import sys
from pathlib import Path

# Set the project root for notebook execution.
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))
from ml.predict import analyze_ticket
```

```python
# Run representative ticket checks.
test_cases = [
    ("Payment failed", "I am trying to make a payment but my card keeps getting declined."),
    ("App crashes on login", "The application crashes every time I open the dashboard."),
    ("Unauthorized transaction on my card", "I noticed unknown charges that I did not authorize."),
]
for number, (subject, description) in enumerate(test_cases, start=1):
    result = analyze_ticket(subject, description)
    print(f"Test case {number}: {subject}")
    print("Category:", result["category"])
    print("Priority:", result["priority"])
    print("Department:", result["department"])
    print("Similar tickets:")
    for ticket in result["similar_tickets"]:
        print(f"  {ticket['Ticket_ID']} - {ticket['Subject']} ({ticket['Similarity']}%)")
    print()
```
